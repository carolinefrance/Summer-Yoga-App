This File Extracts the Data

Mindbody emails are highly templated, which makes them perfect candidates for Regular Expressions (Regex). As a best practice, we will use Python's named capture groups (?P<name>). This makes our regex pattern self-documenting and outputs a clean dictionary, similar to parsing a JSON object in JavaScript.

**Step 1: The Python Regex Extraction Script**

```python
import re

def parse_mindbody_email(email_text: str) -> dict:
    """
    Parses a Mindbody confirmation email to extract class details using named capture groups.
    """
    # We use re.IGNORECASE to handle slight capitalization changes 
    # We use \s+ to handle variable whitespace (like the double space in your example)
    pattern = re.compile(
        r"reservation for\s+(?P<class_name>.*?)\s+"
        r"with\s+(?P<teacher_name>.*?)\s+"
        r"at\s+(?P<studio_name>.*?)\s+"
        r"on\s+[A-Za-z]+,\s+(?P<date>\d{1,2}/\d{1,2}/\d{4})\.\s+"
        r"Class will start at\s+(?P<time>\d{1,2}:\d{2}[ap]m)\s*\.",
        re.IGNORECASE | re.DOTALL
    )

    match = pattern.search(email_text)

    if match:
        # groupdict() automatically maps the ?P<name> tags to dictionary keys
        return match.groupdict()
    else:
        return {"error": "Regex match failed. Email format may have changed."}

# --- Testing the Function ---

sample_email = """
Dear Caroline,
 
This is a reminder for your reservation for  Anniversary $5 Power Flow with Wednesday Cooper at Southern Om Hot Yoga Southern Om West End on Friday, 6/12/2026. Class will start at 6:00am .
"""

extracted_data = parse_mindbody_email(sample_email)

# In Python, using an 'f-string' is similar to JS template literals (`${var}`)
if "error" not in extracted_data:
    print("Extraction Successful!")
    print(f"Teacher: {extracted_data['teacher_name']}")
    print(f"Class:   {extracted_data['class_name']}")
    print(f"Date:    {extracted_data['date']}")
    print(f"Time:    {extracted_data['time']}")
    print(f"Studio:  {extracted_data['studio_name']}")
else:
    print(extracted_data["error"])
```
